# Practical 1A – Feature Transformation using PCA

## Aim
Apply Principal Component Analysis to the Wine dataset and reduce dimensionality while retaining as much variation as possible.

## One-minute viva answer
The Wine dataset has 178 samples, 13 numerical chemical features and 3 wine classes. PCA is an **unsupervised dimensionality-reduction** technique. We standardize the 13 features, find new perpendicular directions called principal components, and retain two components for a 2D representation. The explained-variance ratio tells us how much of the original variation is retained. The class labels can colour the scatter plot, but are not used to fit PCA.

## Theory and terminology
- **Dimensionality:** number of input features; here initially 13.
- **Variance:** spread of values in a feature.
- **Covariance:** how two features vary together.
- **Principal component:** a weighted linear combination of original features.
- **PC1:** direction of maximum projected variance.
- **PC2:** direction of maximum remaining variance, perpendicular to PC1.
- **Eigenvector:** direction of a principal component.
- **Eigenvalue:** amount of variance along that direction.
- **Explained variance ratio:** fraction of total variance represented by a component.
- **Loadings:** weights showing each original feature's contribution.

## Why standardize?
PCA is variance-based. A feature measured in large numerical units can dominate features with smaller scales. Standardization transforms each feature to mean 0 and standard deviation 1:
\[
z=(x-\mu)/\sigma
\]
where x is the value, μ the feature mean and σ its standard deviation.

## PCA algorithm
1. Arrange observations in rows and features in columns.
2. Center/standardize the features.
3. Calculate the covariance structure (or use an equivalent SVD procedure).
4. Find eigenvectors and eigenvalues.
5. Sort components by decreasing eigenvalue.
6. Select the first k components.
7. Project observations onto those components.
8. Inspect explained variance and visualize the result.

## Practical code/output meaning
- Load Wine data; separate the first-column class label from the 13 input features.
- Fit `StandardScaler` to feature data and transform it.
- Fit `PCA(n_components=2)` and transform the standardized data.
- Output shape changes from 13 columns to 2 columns.
- `explained_variance_ratio_` reports the fraction represented by PC1 and PC2.
- The sum of those ratios is the total variation retained in 2D.
- A scatter plot uses PC1 and PC2 as axes and colour for wine class. Clear colour separation suggests the projection preserves class-related structure; overlap means two components do not fully separate classes.

## Advantages / limitations
**Advantages:** fewer features, 2D/3D visualization, reduced redundancy from correlated variables, possible noise reduction and faster downstream learning.
**Limitations:** components can be hard to interpret; PCA captures linear variance; discarded components lose information; high variance is not necessarily predictive; scaling and outliers affect results.

## Viva questions with answers

1. **Question: What is PCA?**
   **Answer:** A technique that transforms original features into fewer uncorrelated components while retaining maximum variance.
2. **Question: Supervised or unsupervised?**
   **Answer:** Unsupervised; labels are not used to find components.
3. **Question: Why standardize?**
   **Answer:** To prevent large-scale features dominating variance.
4. **Question: What is PC1?**
   **Answer:** Direction with maximum variance.
5. **Question: Are PC1 and PC2 correlated?**
   **Answer:** They are orthogonal and uncorrelated in the PCA representation.
6. **Question: What does explained variance ratio mean?**
   **Answer:** Fraction of total variance explained by a component.
7. **Question: Why two components?**
   **Answer:** For 2D visualization and dimensionality reduction from 13 to 2.
8. **Question: PCA vs feature selection?**
   **Answer:** Selection keeps existing columns; PCA creates new combinations.
9. **Question: Does PCA guarantee better accuracy?**
   **Answer:** No; it preserves variance, not necessarily predictive information.
10. **Question: What is a loading?**
   **Answer:** A coefficient/weight linking an original feature to a component.

# Practical 2A – Uber Fare Prediction using Regression

## Aim
Predict a numerical Uber trip fare from historical trip information.

## One-minute viva answer
Uber fare prediction is a **supervised regression** task because the target, fare amount, is continuous. Historical trips have known fares. We clean invalid records, create useful inputs such as trip distance and pickup time, split the data into training and testing sets, train a regression model, predict unseen fares and evaluate errors using MAE, MSE, RMSE and R-squared.

## Regression theory
Regression predicts a continuous number, unlike classification which predicts a category. Simple linear regression:
\[
\hat y=b_0+b_1x
\]
Multiple linear regression:
\[
\hat y=b_0+b_1x_1+\cdots+b_px_p
\]
The coefficients are commonly estimated by minimizing squared prediction errors (ordinary least squares).

## Common dataset fields
The exact dataset version can vary. Typical fields:
- `fare_amount`: target to predict.
- `pickup_datetime`: pickup date/time.
- `pickup_longitude`, `pickup_latitude`: pickup coordinates.
- `dropoff_longitude`, `dropoff_latitude`: drop-off coordinates.
- `passenger_count`: passenger count.
- `key` or trip ID: identifier, usually not a useful predictive feature.

## Cleaning and feature engineering
Check missing values, negative/impossible fares, invalid coordinates, impossible passenger counts and extreme outliers. Valid latitude is between -90 and 90; longitude is between -180 and 180. Investigate outliers rather than automatically deleting them.

Feature engineering creates useful variables from raw columns:
- Extract pickup hour, day, month, day of week or weekend flag from datetime.
- Calculate trip distance from pickup and drop-off coordinates.
- Avoid identifiers that merely identify a row.

**Haversine distance** estimates great-circle distance on a sphere:
\[
a=\sin^2(\Delta\phi/2)+\cos(\phi_1)\cos(\phi_2)\sin^2(\Delta\lambda/2)
\]
\[
d=2R\arcsin(\sqrt a)
\]
Angles are in radians, R is Earth's mean radius (about 6371 km). Straight-line distance is not the same as road distance.

## Models
- **Linear Regression:** fits a linear relationship; simple and interpretable.
- **Decision Tree Regression:** splits feature space into regions and predicts a value in each region.
- **Random Forest Regression:** averages many decision-tree predictions.
- **Gradient Boosting Regression:** builds trees sequentially to correct earlier errors.
Use the algorithm specified in your practical as the main answer.

## Train/test split and leakage
Training data fits the model; testing data estimates performance on unseen records. A common split is 80/20. Fit preprocessing only on training data, then apply it to test data. **Data leakage** occurs when information unavailable at prediction time or test-set information influences training.

## Evaluation metrics
Let yᵢ be actual fare, ŷᵢ predicted fare, n the number of test records.

**MAE:** average absolute error.
\[
MAE=\frac1n\sum_i|y_i-\hat y_i|
\]
It is in fare units and easy to explain.

**MSE:** average squared error.
\[
MSE=\frac1n\sum_i(y_i-\hat y_i)^2
\]
Large errors receive greater penalty; units are squared.

**RMSE:** square root of MSE; returns to fare units and penalizes large errors.
\[
RMSE=\sqrt{MSE}
\]

**R-squared:** compares model squared error with the mean-prediction baseline.
\[
R^2=1-\frac{\sum_i(y_i-\hat y_i)^2}{\sum_i(y_i-\bar y)^2}
\]
Near 1 indicates strong fit; 0 is like predicting the mean; a negative test score means worse than that baseline.

## Reading outputs
- Dataset `shape`, `info`, `describe`: size, data types, missingness and statistics.
- Correlation: linear association, not causation.
- Actual-vs-predicted plot: points near diagonal indicate good predictions.
- Residual = actual − predicted; patterns can indicate model problems.
- Lower MAE/RMSE is generally better; interpret errors in fare units.
- R² is not the percentage of predictions that are correct.

## Limitations
Fare may depend on traffic, route, tolls, surge pricing and other unavailable variables. Coordinates can be erroneous; straight-line distance differs from road distance; outliers and changing pricing patterns affect accuracy. A simple linear model may miss non-linear effects.

## Viva questions with answers

1. **Question: What is regression?**
   **Answer:** Supervised prediction of a continuous numerical value.
2. **Question: Why is fare prediction regression?**
   **Answer:** Fare is numeric and continuous.
3. **Question: Dependent variable?**
   **Answer:** Usually `fare_amount`.
4. **Question: Independent variables?**
   **Answer:** Trip distance, coordinates, time, passenger count, etc.
5. **Question: Classification vs regression?**
   **Answer:** Category vs continuous number.
6. **Question: Why train/test split?**
   **Answer:** To evaluate generalization to unseen data.
7. **Question: Overfitting?**
   **Answer:** Good training performance but poor unseen-data performance.
8. **Question: MAE?**
   **Answer:** Mean absolute difference between actual and predicted fares.
9. **Question: Why RMSE?**
   **Answer:** Fare units, with stronger penalty for large errors.
10. **Question: R²?**
   **Answer:** Improvement relative to predicting the target mean; may be negative on test data.
11. **Question: Residual?**
   **Answer:** Actual minus predicted value.
12. **Question: Feature engineering?**
   **Answer:** Creating useful inputs, e.g. pickup hour and trip distance.
13. **Question: Why calculate distance?**
   **Answer:** It is often related to fare and more useful than raw coordinates alone.
14. **Question: What is data leakage?**
   **Answer:** Training with information that should not be available or that comes from test data.
15. **Question: Does high R² guarantee a good model?**
   **Answer:** No; check test errors, residuals and data quality.

# Practical 3A – Handwritten Digit Classification using SVM

## Aim
Classify handwritten digit images into labels 0 through 9 using Support Vector Machine.

## One-minute viva answer
The scikit-learn Digits dataset contains 8×8 grayscale images. Each image can be represented by 64 pixel-intensity features, and its label is a digit from 0 to 9. We split labelled examples into training and testing sets, train an SVM classifier, predict test labels and evaluate using accuracy and a confusion matrix.

## SVM theory
Support Vector Machine is a supervised algorithm for classification and regression. In classification it finds a decision boundary, or **hyperplane**, that separates classes while maximizing the **margin** to the nearest training points.

- **Support vectors:** closest training points that determine the boundary.
- **Margin:** distance from boundary to nearest points.
- **Kernel:** enables non-linear decision boundaries using feature-space relationships.
- **C:** controls trade-off between wide margin and training errors. Higher C penalizes errors more strongly and may overfit.
- **Gamma:** for RBF kernel, controls influence range of each training example.

Common kernels:
- **Linear:** linear decision boundary.
- **Polynomial:** polynomial relationships.
- **RBF/Gaussian:** flexible non-linear boundary.
- **Sigmoid:** sigmoid-like kernel, less commonly used.

For multi-class SVM, multiple binary classifiers are combined. Scikit-learn SVC uses one-vs-one internally.

## Workflow and output
1. Load Digits data and inspect example images.
2. Each 8×8 image corresponds to 64 pixel values (the `data` array is already flattened).
3. Split into training/testing sets; stratification preserves class proportions.
4. Train SVM; predict test labels.
5. Accuracy = correct predictions / total predictions.
6. A 10×10 confusion matrix compares actual and predicted digit classes. In scikit-learn, rows are actual and columns predicted. Diagonal = correct; off-diagonal = mistakes.
Misclassifications may occur when handwriting styles look similar.

## Advantages / limitations
**Advantages:** effective in high-dimensional spaces, kernels handle non-linear boundaries, decision relies on support vectors.
**Limitations:** can be slow on large datasets; kernel/hyperparameter choice matters; scaling is often important; less interpretable than simple linear models.

   **Answer:** A decision boundary separating classes.
4. **Question: Margin?**
   **Answer:** Distance to the nearest training points.
5. **Question: Why kernel?**
   **Answer:** To model non-linear boundaries.
6. **Question: RBF kernel?**
   **Answer:** Gaussian kernel for flexible non-linear boundaries.
7. **Question: What does C do?**
   **Answer:** Controls penalty for training errors.
8. **Question: What does gamma do?**
   **Answer:** Controls influence range in RBF SVM.

   **Answer:** 64 (8×8 pixels).

   **Answer:** 10, digits 0–9.

# Practical 4A – K-Means Clustering on Iris using Elbow Method

## Aim
Cluster Iris observations using K-Means and use the Elbow Method to help choose K.

## One-minute viva answer
K-Means is an **unsupervised clustering** algorithm. It assigns observations to K groups based on distance to cluster centroids. It repeatedly assigns each point to its nearest centroid and recalculates centroids as the mean of assigned points. The Elbow Method plots within-cluster sum of squares (WCSS/inertia) against K and looks for the bend where adding clusters gives diminishing improvement.

## Dataset and theory
The common Iris dataset has 150 samples, four numerical measurements (sepal length, sepal width, petal length, petal width) and three known species. K-Means must be fitted using measurements, not species labels. Labels may be compared afterwards; cluster IDs are arbitrary and do not automatically equal species names.

**Euclidean distance:**
\[
d(x,c)=\sqrt{\sum_{j=1}^{p}(x_j-c_j)^2}
\]
**Centroid:** mean feature vector of points in a cluster.

**WCSS / inertia:**
\[
WCSS=\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2
\]
K-Means minimizes this within-cluster squared distance.

## K-Means steps
1. Choose K.
2. Initialize K centroids (K-Means++ is a common initialization).
3. Assign each point to its nearest centroid.
4. Recalculate each centroid as the mean of assigned points.
5. Repeat until assignments/centroids stabilize or iteration limit is reached.

## Elbow Method
Fit for several K values, record inertia, plot K vs inertia and find the bend where the decline becomes less steep. WCSS always decreases or stays the same as K grows, so simply choosing the lowest WCSS would favour too many clusters. The elbow is a heuristic, not a guaranteed unique answer.

## Scaling and outputs
K-Means uses distances. Features with larger ranges can dominate, so standardization is often useful. The elbow graph shows inertia vs K; cluster labels assign each observation a group; centroids show cluster centres; scatter plots visualize assignments using selected features or a 2D projection.

## Advantages / limitations
**Advantages:** simple, fast, easy to implement, works well for compact roughly spherical clusters.
**Limitations:** K must be chosen; sensitive to initialization and outliers; can reach local optima; weak for irregular shapes or clusters with very different density/size.

## Viva questions with answers

1. **Question: What is K-Means?**
   **Answer:** Unsupervised partitioning into K groups around centroids.
2. **Question: Supervised?**
   **Answer:** No, labels are not required.
3. **Question: Centroid?**
   **Answer:** Mean position of points in a cluster.
4. **Question: WCSS?**
   **Answer:** Sum of squared distances to assigned centroids.
5. **Question: Why does WCSS fall as K increases?**
   **Answer:** More centroids allow closer assignments.
6. **Question: Elbow Method?**
   **Answer:** Heuristic choosing K at diminishing returns in WCSS.
7. **Question: Why scale?**
   **Answer:** Distance can be dominated by large-range features.
8. **Question: Does cluster 0 mean Setosa?**
   **Answer:** No, cluster numbers are arbitrary.
9. **Question: K-Means++?**
   **Answer:** Initialization that spreads centroids apart.
10. **Question: K-Means limitation?**
   **Answer:** Sensitive to outliers/initialization and assumes compact clusters.
11. **Question: K-Means vs K-Medoids?**
   **Answer:** K-Means uses a mean centroid; K-Medoids uses an actual observation as representative.

# Practical 5A – Random Forest for Car Safety Prediction

## Aim
Predict car safety category from car attributes using a Random Forest classifier.

## One-minute viva answer
Random Forest is a supervised ensemble classifier. It builds many decision trees using bootstrap samples and random subsets of features. Each tree predicts a class and the forest selects the majority-vote class. The common Car Evaluation data has buying, maintenance, doors, persons, luggage boot and safety attributes. For this practical, `safety` is the target and should not be included among input features. The overall `class`/acceptability field should also be excluded to avoid leakage or changing the task.

## Dataset and decision trees
Common fields:
- `buying`: buying-price category.
- `maint`: maintenance-cost category.
- `doors`: door count.
- `persons`: passenger capacity.
- `lug_boot`: luggage boot size.
- `safety`: safety category (target here).
- `class`: overall acceptability category.

A decision tree asks feature-based questions. Internal nodes test conditions, branches represent outcomes and leaves predict a class.

**Gini impurity:**
\[
Gini=1-\sum_{i=1}^{C}p_i^2
\]
**Entropy:**
\[
H=-\sum_{i=1}^{C}p_i\log_2p_i
\]
Both measure class mixture/uncertainty in a node; good splits reduce impurity.

## Random Forest algorithm
1. Draw a bootstrap sample (sample with replacement) for each tree.
2. At each split, consider a random subset of features.
3. Train a tree.
4. Repeat to create many diverse trees.
5. For classification, use majority voting across trees.

Random feature selection reduces correlation among trees. Combining many diverse trees usually reduces variance compared with a single deep tree.

## Encoding, leakage and evaluation
Categorical values must be encoded, commonly with one-hot encoding (a category becomes binary indicator columns). A preprocessing pipeline ensures training-fitted encoding is also applied consistently to test data.

**Target leakage warning:** if predicting `safety`, `safety` is the target, not an input. The overall acceptability `class` is closely related and must not be used as an input for this task.

Metrics:
- Accuracy = correct predictions / total predictions.
- Precision = TP/(TP+FP): of predicted class members, how many are correct.
- Recall = TP/(TP+FN): of actual class members, how many are found.
- F1 = harmonic mean of precision and recall.
- Confusion matrix: actual vs predicted counts; diagonal entries are correct.
Macro averaging gives equal weight to each class; weighted averaging accounts for class support.

## Outputs
Accuracy is overall test correctness. A classification report shows precision, recall, F1 and support per class. Confusion matrix exposes which classes are confused. An example prediction shows the predicted safety class for a new attribute combination. Feature importance estimates contribution to tree impurity reductions; it is not proof of causation.

## Advantages / limitations
**Advantages:** models non-linear relations/interactions, often more stable than one tree, handles many features, provides feature-importance estimates.
**Limitations:** less interpretable, more memory/computation, importance measures have caveats, cannot repair biased or poor-quality data.

   **Answer:** Ensemble of randomized decision trees.
2. **Question: Supervised?**
   **Answer:** Yes, trained with known safety labels.
3. **Question: Why forest?**
   **Answer:** Many trees contribute predictions.
4. **Question: Bootstrap?**
   **Answer:** Sampling with replacement.
5. **Question: How classify?**
   **Answer:** Majority vote.
6. **Question: Why random features?**
   **Answer:** Increase diversity and reduce tree correlation.
7. **Question: Gini impurity?**
   **Answer:** Measure of class mixing in a node.
8. **Question: One-hot encoding?**
   **Answer:** Binary indicator columns for categories.
9. **Question: Why exclude `class`?**
   **Answer:** It can leak overall acceptability information and change the intended safety prediction task.
10. **Question: Feature importance?**
   **Answer:** Estimate of feature contribution to splits, not causation.
11. **Question: Decision Tree vs Random Forest?**
   **Answer:** One tree vs ensemble of many trees.
12. **Question: What is overfitting?**
   **Answer:** Training-specific learning that does not generalize.

# Practical 6C – Tic-Tac-Toe using Q-Learning

## Aim
Train a Tic-Tac-Toe agent using Q-Learning, learning from rewards and penalties.

## One-minute viva answer
Reinforcement Learning is a method where an agent interacts with an environment and receives rewards or penalties. In Tic-Tac-Toe, the board is the state, placing a mark in an empty cell is an action, and the final outcome produces a reward. Q-Learning learns values for state-action pairs. During training the agent explores moves; later it can choose the legal move with the highest learned Q-value. In this practical, the X agent trains and is tested against a random O opponent, with wins, draws and losses counted.

## RL components
- **Agent:** learner/decision-maker (the game-playing AI).
- **Environment:** Tic-Tac-Toe game and rules.
- **State:** current board configuration (and turn information if needed).
- **Action:** placing a mark in an empty square.
- **Reward:** numerical feedback for an outcome or transition.
- **Policy:** strategy for selecting actions.
- **Episode:** one complete game.
- **Terminal state:** game-over state (win, loss or draw).

## Q-Learning
Q-Learning is model-free and off-policy. It learns a value Q(s,a) for taking action a in state s.

\[
Q(s,a)\leftarrow Q(s,a)+\alpha[r+\gamma\max_{a'}Q(s',a')-Q(s,a)]
\]

- Q(s,a): current action-value estimate.
- α: learning rate; how strongly new information changes the old estimate.
- r: immediate reward.
- γ: discount factor; importance of future reward.
- s′: next state.
- max Q(s′,a′): best estimated future value in next state.
For terminal states, future value is zero.

## Learning process
1. Initialize Q-values/Q-table.
2. Start an empty game.
3. Observe the current state.
4. Select a legal action.
5. Apply the move and observe reward and next state.
6. Update Q-value using the formula.
7. Continue until the game ends.
8. Repeat over many episodes.
9. Evaluate the learned policy, often selecting the highest-Q legal action.

## Exploration vs exploitation
- **Exploration:** try actions to learn their consequences.
- **Exploitation:** choose the best-known action.
- **Epsilon-greedy:** explore randomly with probability ε; otherwise exploit.
- **Epsilon decay:** reduce exploration gradually so the agent explores early and exploits more later.

A typical reward design gives positive reward for a win, negative for a loss and zero or small reward for a draw. Exact reward numbers are implementation choices.

## Training and testing
Training updates Q-values over many games. Testing evaluates the learned policy without relying on continued random exploration. Win rate = wins / test games; draw rate = draws / test games; loss rate = losses / test games. These metrics are meaningful only with the opponent and test conditions stated. A high score against a random opponent does not prove unbeatable play.

## Why model-free and off-policy?
Model-free: the agent learns from interaction rather than requiring a complete transition model in advance. Off-policy: the update uses the maximum next-state Q-value, even if the behaviour policy used an exploratory action.

## Outputs
Q-values represent estimated long-term action values. A training reward curve can show learning trends but may be noisy. Win/draw/loss counts summarize test performance. A demonstration game shows the selected moves. A simple setup does not guarantee an optimal strategy.

## Advantages / limitations
**Advantages:** learns from experience, no hand-coded complete strategy needed, suitable for sequential decisions, simple update.
**Limitations:** tabular state space can grow; many episodes may be needed; reward design and exploration matter; performance varies with random seed and opponent.

   **Answer:** Estimated long-term return for an action in a state.
4. **Question: Learning rate α?**
   **Answer:** Controls update size.
5. **Question: Discount γ?**
   **Answer:** Controls importance of future rewards.

   **Answer:** Random exploration with probability ε, otherwise best-known action.

- PCA does not use class labels to find components.
- High PCA variance does not guarantee predictive usefulness.
- Regression predicts a number, not a class.

- Random Forest is an ensemble, not one large tree.
- Do not include the target or leaked information as an input.
- Q-Learning learns from reward feedback, not labels specifying the correct move.
- RL results depend on opponent, reward design, exploration and test conditions.
