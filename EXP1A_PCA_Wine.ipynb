# EXPERIMENT 1A – Feature Transformation using PCA
## Aim
To apply Principal Component Analysis (PCA) to the Wine dataset and reduce dimensionality while retaining most of the variation in the measurements.

## 1. Import required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## 2. Load the dataset

```python
# Read the Wine dataset directly from the given URL
url = "https://media.geeksforgeeks.org/wp-content/uploads/Wine.csv"
df = pd.read_csv(url)

# Display the first five records and dataset shape
display(df.head())
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

## 3. Separate features and wine type

```python
# In this dataset, the first column is the wine type/target.
# Remaining columns contain numerical measurements.
target = df.iloc[:, 0]
X = df.iloc[:, 1:]

print("Wine classes:", target.unique())
print("Number of features:", X.shape[1])
```

## 4. Standardize the features

```python
# PCA is affected by different feature scales.
# StandardScaler transforms each feature to mean 0 and standard deviation 1.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 5. Apply PCA

```python
# Reduce the dataset to two principal components for visualization.
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame containing the two principal components.
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Wine_Type"] = target.values

print("Transformed data (first five rows):")
display(pca_df.head())

# Variance explained by each principal component
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance retained:", pca.explained_variance_ratio_.sum())
```

## 6. Visualize principal components

```python
# Plot PC1 against PC2 and use different colors for wine types.
plt.figure(figsize=(8, 6))
for wine_type in pca_df["Wine_Type"].unique():
    subset = pca_df[pca_df["Wine_Type"] == wine_type]
    plt.scatter(subset["PC1"], subset["PC2"], label=str(wine_type), alpha=0.7)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Wine dataset (2 principal components)")
plt.legend(title="Wine Type")
plt.grid(True)
plt.show()
```

## Result / Conclusion
PCA transformed the standardized wine measurements into two principal components. The explained variance ratio indicates how much information is retained by these components. The scatter plot helps visualize whether wine types form separate groups in the reduced two-dimensional space.
