# EXPERIMENT 1B – Classification using LDA
## Aim
To apply Linear Discriminant Analysis (LDA) on the Iris dataset and classify a flower into its species.

## 1. Import required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 2. Load the Iris dataset

```python
# Load Iris dataset (150 samples, 4 features, 3 species)
iris = load_iris()

X = iris.data                 # Sepal/petal measurements
y = iris.target                # Numeric species labels: 0, 1, 2
species_names = iris.target_names

# Show data in tabular form
df = pd.DataFrame(X, columns=iris.feature_names)
df["Species"] = [species_names[i] for i in y]
display(df.head())
print("Dataset shape:", df.shape)
print("Species:", species_names)
```

## 3. Split data into training and testing sets

```python
# Use 80% records for training and 20% for testing.
# stratify=y keeps all three species represented proportionally.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 4. Train the LDA classifier

```python
# Create and train the Linear Discriminant Analysis model.
lda = LinearDiscriminantAnalysis()
lda.fit(X_train, y_train)
```

## 5. Predict species and evaluate the model

```python
# Predict species labels for the test data.
y_pred = lda.predict(X_test)

# Calculate model accuracy and display evaluation details.
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=species_names))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

## 6. Classify a new flower

```python
# Example flower measurements in cm:
# [sepal length, sepal width, petal length, petal width]
new_flower = [[5.1, 3.5, 1.4, 0.2]]

# Predict numeric class and convert it to the species name.
predicted_class = lda.predict(new_flower)[0]
predicted_species = species_names[predicted_class]

print("Predicted species:", predicted_species)
```

## Result / Conclusion
The LDA model was trained using Iris flower measurements and tested on unseen samples. It predicts the species of a flower from its sepal and petal measurements. The accuracy and confusion matrix summarize the classification performance.
