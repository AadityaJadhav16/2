# EXPERIMENT 3A – Classification Analysis using Support Vector Machine (SVM)

**Aim:** To implement a Support Vector Machine classifier to classify images of handwritten digits into numerical classes from 0 to 9.

**Dataset:** Scikit-learn's built-in handwritten digits dataset (`load_digits`). It contains 8 × 8 pixel images of handwritten digits and does not require a separate download.

**Algorithm:** Support Vector Machine (SVM) with an RBF kernel.

## 1. Import required libraries

We use NumPy for numerical operations, Matplotlib to display digit images, and Scikit-learn for the dataset, preprocessing, model training and evaluation.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Load the handwritten digits dataset
from sklearn.datasets import load_digits

# Tools for splitting data and scaling pixel values
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Support Vector Machine classifier
from sklearn.svm import SVC

# Evaluation metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 2. Load and explore the dataset
The dataset contains images represented as arrays of pixel-intensity values. The target contains the correct digit (0 to 9).

```python
# Load the built-in handwritten digits dataset
digits = load_digits()

# X contains flattened pixel values; y contains digit labels (0 to 9)
X = digits.data
y = digits.target

# Display basic dataset information
print("Number of images:", X.shape[0])
print("Number of features per image:", X.shape[1])
print("Digit classes:", np.unique(y))
```

## 3. Display sample handwritten digit images
Each image is 8 × 8 pixels. The label above each image is its actual digit.

```python
# Display the first 10 digit images
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(digits.images[i], cmap="gray")
    plt.title("Label: " + str(y[i]))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## 4. Split the dataset into training and testing sets
We use 80% of the images for training and 20% for testing. `stratify=y` keeps the class proportions similar in both sets.

```python
# Split data; random_state makes the split reproducible
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 5. Feature scaling
Scaling puts pixel features on a comparable scale. The scaler is fitted only on training data to avoid data leakage.

```python
# Create the scaler and learn scaling values from training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling transformation to test data
X_test_scaled = scaler.transform(X_test)
```

## 6. Create and train the SVM classifier
The RBF kernel can learn non-linear decision boundaries. The classifier learns the relationship between pixel values and digit labels.

```python
# Create an SVM classifier using the Radial Basis Function kernel
svm_model = SVC(kernel="rbf", gamma="scale")

# Train the model using the scaled training data
svm_model.fit(X_train_scaled, y_train)

print("SVM model training completed.")
```

## 7. Predict digit classes
The trained model predicts the digit for every image in the test set.

```python
# Predict labels for the unseen test images
y_pred = svm_model.predict(X_test_scaled)

# Show a few actual and predicted labels
print("Actual labels:   ", y_test[:15])
print("Predicted labels:", y_pred[:15])
```

## 8. Evaluate the model
The confusion matrix compares actual and predicted classes. Accuracy is the proportion of correct predictions. The classification report gives precision, recall and F1-score for each digit.

```python
# Calculate overall accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", round(accuracy * 100, 2), "%")

# Display confusion matrix (rows = actual, columns = predicted)
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:\n", cm)

# Display precision, recall and F1-score for digits 0 to 9
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

## 9. Visualize some test predictions
The title shows the predicted digit. Correct predictions are marked as correct; incorrect predictions show both values.

```python
# Display the first 10 test images with their predictions
plt.figure(figsize=(10, 4))

for i in range(10):
    # Reshape the flattened 64 features back into an 8 x 8 image
    image = X_test[i].reshape(8, 8)
    plt.subplot(2, 5, i + 1)
    plt.imshow(image, cmap="gray")
    plt.title(f"Actual: {y_test[i]}\nPredicted: {y_pred[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## 10. Conclusion

The Support Vector Machine classifier was successfully implemented on the handwritten digits dataset. The model was trained to recognize handwritten digits from 0 to 9, and its performance was evaluated using accuracy, a confusion matrix, precision and recall.

**Practical note:** This experiment uses Scikit-learn's small built-in 8 × 8 digits dataset so it runs directly in Jupyter Notebook without downloading MNIST.
